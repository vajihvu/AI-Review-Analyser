# AI Customer Review Analyzer

This notebook provides a complete walk-through of the end-to-end NLP system for analyzing customer reviews. The pipeline processes review datasets and performs:
1. **Data Preprocessing & Cleaning**
2. **Transformer-based Sentiment Analysis** (TensorFlow backend)
3. **Topic Modeling with LDA** (GenSim)
4. **Abstractive Summarization** (TensorFlow backend)

### 1. Imports and Setup

```python
import os
import pandas as pd
from src.preprocess import preprocess
from src.sentiment import analyze_sentiment
from src.topics import topic_modeling
from src.summarize import summarize_reviews
```

### 2. Run Data Preprocessing
This step loads the raw reviews dataset, extracts relevant fields, filters out HTML markup and punctuation, and outputs a cleaned text column.

```python
RAW_DATA = 'data/raw/reviews.csv'
PROCESSED_DATA = 'data/processed/cleaned_reviews.csv'

print("Running preprocessing...")
df = preprocess(RAW_DATA, PROCESSED_DATA)
print(f"Preprocessed {len(df)} reviews successfully!")
df.head(5)
```

### 3. Sentiment Analysis (TensorFlow Backend)
We use a pre-trained DistilBERT model loaded through the Hugging Face `transformers` library on top of TensorFlow to predict the sentiment of the reviews.

```python
sample_size = 20
sample_reviews = df.loc[:sample_size-1, "cleaned_review"].tolist()

print("Analyzing sentiment using TensorFlow...")
sentiments = analyze_sentiment(sample_reviews)

for review, sentiment in zip(df.loc[:5, "review_text"].tolist(), sentiments[:6]):
    print(f"Review: {review[:100]}...")
    print(f"Sentiment: {sentiment}\n")
```

### 4. Topic Modeling with LDA
We use GenSim's Latent Dirichlet Allocation (LDA) to extract the most descriptive topics from the reviews, filtering out generic review filler words.

```python
import re

print("Extracting topics...")
topics = topic_modeling(df["cleaned_review"].dropna().tolist()[:500])

print("\nExtracted Topics:")
for idx, topic_str in topics:
    words = re.findall(r'"([^"]+)"', topic_str)
    words = [w for w in words if w != "br"][:5]
    print(f"Topic #{idx + 1}: {', '.join(words)}")
```

### 5. Abstractive Summarization (TensorFlow Backend)
We use the BART Large CNN model configured with the TensorFlow backend to summarize the core feedback of a collection of reviews into a single paragraph.

```python
print("Generating summary using TensorFlow...")
combined_text = " ".join(df["review_text"].tolist()[:20])
summary = summarize_reviews(combined_text)
print(f"\nSummary Paragraph:\n{summary}")
```
